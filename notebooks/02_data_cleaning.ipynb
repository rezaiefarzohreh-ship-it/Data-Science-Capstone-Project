```python
import pandas as pd
```

```python
df_raw = pd.read_excel("../data/raw/Concrete_Data.xls")
```

```python
df_clean = df_raw.copy()
```

```python
df_clean.columns = [
    'cement',
    'slag',
    'fly_ash',
    'water',
    'superplasticizer',
    'coarse_aggregate',
    'fine_aggregate',
    'age',
    'strength'
]
```

```python
df_clean.head()
```

```python
df_clean.columns
```

```python
df_clean.duplicated().sum()
```

```python
df_clean = df_clean.drop_duplicates().reset_index(drop=True)
```

```python
df_clean.shape
```

```python
df_clean.duplicated().sum()
```

```python
material_cols = [
    'cement',
    'slag',
    'fly_ash',
    'water',
    'superplasticizer',
    'coarse_aggregate',
    'fine_aggregate'
]
```

```python
df_clean['mix_id'] = df_clean.groupby(material_cols, sort=False).ngroup()
```

```python
df_clean.head()
```

```python
df_clean['mix_id'].nunique()
```

```python
def assign_age_stage(age):
    if 1 <= age <= 14:
        return 'Early'
    elif age == 28:
        return '28-day'
    elif 56 <= age <= 365:
        return 'Later'
    else:
        return 'Other'
```

```python
df_clean['age_stage'] = df_clean['age'].apply(assign_age_stage)
```

```python
df_clean[['age', 'age_stage']].head(20)
```

```python
df_clean['age_stage'].value_counts()
```

```python
(df_clean['age_stage'] == 'Other').sum()
```

```python
ages_per_mix = df_clean.groupby('mix_id')['age'].nunique()

(ages_per_mix > 1).sum()
```
